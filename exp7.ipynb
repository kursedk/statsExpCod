# Pima Indians Diabetes Prediction

First, let's load the Pima Indians Diabetes Dataset. We will download it from the UCI Machine Learning Repository, as it's a commonly used source for this dataset, and then load it into a pandas DataFrame.

```python
import pandas as pd

# URL for the Pima Indians Diabetes Dataset from UCI Machine Learning Repository
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.csv"

# Define column names based on the dataset description
# This dataset typically doesn't have a header row, so we need to provide names
column_names = [
    'pregnancies',
    'glucose',
    'blood_pressure',
    'skin_thickness',
    'insulin',
    'bmi',
    'diabetes_pedigree_function',
    'age',
    'outcome'
]

# Load the dataset
df = pd.read_csv(url, names=column_names)

# Display the first 5 rows of the DataFrame
display(df.head())
```

## 2. Separate input features and target variable

We will now separate the features (all columns except 'outcome') from the target variable ('outcome').

```python
# Separate input features (X) and target variable (y)
X = df.drop('outcome', axis=1)
y = df['outcome']

print("Shape of features (X):", X.shape)
print("Shape of target (y):", y.shape)
display(X.head())
display(y.head())
```
